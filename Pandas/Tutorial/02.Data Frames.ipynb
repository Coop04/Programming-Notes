```python
people = {
    "first": ["Corey"], 
    "last": ["Schafer"], 
    "email": ["CoreyMSchafer@gmail.com"]
}
```

```python
people = {
    "first": ["Corey", 'Jane', 'John'], 
    "last": ["Schafer", 'Doe', 'Doe'], 
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
people.get("email")
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people) # converts a dictionary into Data Frame
```

```python
df # Data Frame is a 2-D array
```

Accessing Columns

```python
df['email'] # Gives a series (1-D array) of email along with an index
```

```python
df.email # dot notation can also be used (unless the variable name is not a method name like count)
```

```python
type(df['email'])  # shows the type (here series means a 1-D array)
```

```python
# to access multiple locations
df[['last','email']]
```

```python
type(df[['last','email']])
```

```python
# to access columns
df.columns
```

Accessing rows and columns using loc & iloc

```python
# to access rows we use loc and iloc (integer location)
df.iloc[0] # gives the 0th row
```

```python
# to access multiple rows
df.iloc[[0,2]] # returns the 0th and 2nd rows
```

```python
# to access rows along with specific column
df.iloc[[0,1],2] # here 2 is the index for email
```

```python
# to access specific rows and specific columns
df.iloc[[0,1],[1,2]] 
```

```python
df.loc[0]
```

```python
df.loc[[0,1]]
```

```python
df.loc[[0,1],'email'] # in loc we can specify the column by name and not index
```

```python
df.loc[[0,1],['first','email']] # to access mupltiple rows and columns (in the order specified in this code)
```

```python
# to conver series to data frame
df['first'].to_frame()
```

```python
# to convert a data frame to series (squeeze())
df['first'].to_frame().squeeze()
```
