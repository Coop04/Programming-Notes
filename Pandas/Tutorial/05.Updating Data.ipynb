```python
people = {
    "first": ["Corey", 'Jane', 'John'], 
    "last": ["Schafer", 'Doe', 'Doe'], 
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.info()
```

Updating Columns

```python
df.columns
```

```python
# to change the column names
df.columns = ['first name', 'last name', 'email']
df
```

```python
# for lower case; for upper use upper()
df.columns = [x.lower() for x in df.columns]
df.columns
```

```python
# to replace (space with underscore)
df.columns = df.columns.str.replace(' ','_')
df.columns
```

To change only specific columns

```python
# to change the column names make a dictionary with current names as keys and new names as values
df.rename(columns={'first_name':'first','last_name':'last'}, inplace=True)
df
```

Updating the data in the rows

```python
df.loc[2] = df.loc[2].str.replace('Doe','Smith')
df
```

```python
# or make a list of the rows you want to change
df.loc[2]=['John','Smith','JohnSmith@gmail.com']
df
```

```python
# or to only cahnge specific data values
df.loc[2,['last','email']] = ['Dalton','JohnDalton@email.com']
df
```

```python
# to change a single value (no need for a list)
df.loc[2,'last'] = 'Doe'
df
```

```python
# also can use at instead of loc
df.at[2,'last'] = 'Dalton'
df
```

```python
filt = (df['last']=='Dalton')
df[filt]['last'] = 'Doe' # this will give a warning (and not change the value) because it is not clear if you want to change the original df or a copy of it
```

```python
df  # the value is not changed
# ie, when you need to change a value in a df, use loc or at (and not the filter method)
```

```python
# instead of using loc or at, you can use the .loc method with the filter
filt = (df['last']=='Dalton')
df.loc[filt, 'last'] = 'Doe'
df
```

# apply, map, applymap,  replace

### apply: 
running apply on a series applies a function to every value in the series and <br>
running apply to a data frame applies a function to every series in the data frame

running apply on a series

```python
# apply : to apply a function to a series (also used to update values as well)
df['email'].apply(len)
```

```python
def update_email(email):
    return email.upper()   # value must be returned
```

```python
df['email'].apply(update_email) # no need to use paranthesis for the function
```

```python
df['email'] = df['email'].apply(update_email)
df
```

```python
# for shorter functions, use lambda functions
# lammbda functions are anonymous functions (ie, functions without a name)
df['email'] = df['email'].apply(lambda x: x.lower()) # syntax no_name_function : return_value
df 
# lambda functions can aslo be used with numbers to run calculations
```

```python
df['email'].apply(len)
```

running apply to a dataframe

```python
df.apply(len) # this will return the number of rows in each column
# this is because the apply function is applied to each series in the df (ie, each column) is, it says that first has 3 values, last has 3 values, email has 3 values
# (ie, counted verticlly)
```

```python
len(df['email']) # this will return the number of values in the email column
```

```python
# to apply the apply function to each row, use the axis parameter
df.apply(len,axis='columns') # default value of axis is rows
# this will return the number of columns in each row (ie, counted horizontally)
```

```python
# series objects has a min method
df.apply(pd.Series.min) # this will return the min value in each column (ie, here in alphabetical order)
# if used in numerical data, it will return the min value in each column

# we can use extra attribute axis, 
# axis=0 or axis='index' for column wise & 
# axis=1 or axis='columns' for row wise
df.apply(pd.Series.min, axis='index')
```

```python
# same as above
df.apply(lambda x: x.min()) # this x here is a series not a value (and series have a min method)
```

```python
df.applymap(str.lower)
```

```python
df.map(str.lower) 
```

### map:
used for substituting each value in a series with another value (works on a series and data frames) <br>
when used with series: applies a function to each element in a Series (a single column). <br>
when applied with data frames: applies a function to each element in an entire DataFrame (multiple columns).(similar to applymap)


```python
df['first'].map({'Corey':'Chris','Jane':'Mary'}) # this will change only the values that are in the dictionary (ie, it will change Corey to Chris and Jane to Mary) 
#other values (values not substituted) will become NaN
```

### replace:
similar to map but does not give NaN values for data values which wasn't substituted

```python
df['first'].replace({'Corey':'Chris','Jane':'Mary'}) # this will change only the values that are in the dictionary (ie, it will change Corey to Chris and Jane to Mary) other values will remain the same
```

```python
# to set the values in the df to the new values
df['first'] = df['first'].replace({'Corey':'Chris','Jane':'Mary'})
df 
```
