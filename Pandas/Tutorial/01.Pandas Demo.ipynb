### Q: To find percentage of people knowing python in all countries

```python
# to find the percentage of people knowing python in all countries
country_respondents = df['Country'].value_counts()
country_respondents
# country_class['Country'].value_counts()
```

```python
country_uses_python = country_class['LanguageHaveWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
country_uses_python
```

```python
# concatinating total participants from each country with total participants in each country who knows pyhton
python_df = pd.concat([country_respondents,country_uses_python], axis='columns')
python_df
```

```python
# to rename the columns since they are unrelated names
python_df.rename(columns={'count':'country_respondents', 'LanguageHaveWorkedWith':'country_uses_python'},inplace=True)
# python_df.columns = ['country_respondents', 'country_uses_python'] # this method can also be used to rename but all values must be provided
python_df
```

```python
# for calculating the percentage (we add a new column for the percentage values)
python_df['percentagepy']= (country_uses_python/country_respondents)*100
python_df
```

```python
python_df.sort_values(by='percentagepy',ascending=False,inplace=True) # it also works without by=
python_df.head(25)
```

# Cleaning Data

### Q: To calculate the average number of years of coding experience

```python
df['YearsCode'].head(10)
```

```python
# df['YearsCode']=df['YearsCode'].astype(float) 
# # we get an error because of the 'Less than 1 year' value
df['YearsCode'].unique() # to get the unique values in the column
# here we see 'Less than 1 year' and 'More than 50 years' which are string value and cannot be converted to float
```

```python
# so we eill reeplace these values with other values 
# (inplace=True cannot be used since it will be deprecated in the future versions of pandas)
df['YearsCode'] = df['YearsCode'].replace('Less than 1 year',0)
df['YearsCode'] = df['YearsCode'].replace('More than 50 years',51)
df['YearsCode'].unique()
```

```python
# since they are still objects (strings) we need to convert them to float 
df['YearsCode']=df['YearsCode'].astype(float)
```
