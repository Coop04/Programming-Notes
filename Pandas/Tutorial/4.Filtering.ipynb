```python
people = {
    "first": ["Corey", 'Jane', 'John'], 
    "last": ["Schafer", 'Doe', 'Doe'], 
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

Filtering

```python
filt = df['last']=='Doe' # cant use filter as a variable since it is a built in function
df[filt]
```

```python
df[df['last']=='Doe'] # same as above
```

```python
df.loc[filt] # same as above (loc is used to access a group of rows and columns by labels or a boolean array) boolean array is used here
```

```python
df.loc[filt, 'email'] # same as above but only email column (advantage of using loc)
```

Other filtering methods & (and), | (or)

```python
filte = (df['last']=='Doe') & (df['first']=='John') # the & is used to combine two filters which should be in brackets
df.loc[filte]
```

```python
fil = (df['last']=='Schafer') | (df['first']=='John')
df.loc[fil]
```

Opposite of filter ~ (not)

```python
fil = (df['last']=='Schafer') | (df['first']=='John')
df.loc[~fil] # ~ is used to negate the filter (opposite of the filter)
```
