# **Exception Handling in Python**  

## **What is an Exception?**  
An **exception** is an error that occurs **during program execution**, disrupting the normal flow of the program. Unlike syntax errors, exceptions occur when the code is syntactically correct but runs into an issue during execution, such as division by zero or accessing an invalid list index.

---

## **How to Handle Exceptions in Python?**  
Python provides a structured way to handle exceptions using the **`try-except`** block. The `try` block contains the code that may cause an exception, and the `except` block catches and handles that exception.

### **Basic Syntax**
```python
try:
    # Code that might raise an exception
    risky_code()
except ExceptionType:
    # Code to handle the exception
```

---

## **Types of Exceptions with Examples**

### **1. `ZeroDivisionError` (Division by Zero)**
Occurs when a number is divided by zero.

```python
try:
    num = 10 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```
### **Output:**
```
Error: Cannot divide by zero.
```

---

### **2. `IndexError` (Invalid Index in a List)**
Occurs when accessing an index that does not exist in a list.

```python
try:
    my_list = [1, 2, 3]
    print(my_list[5])  # Index out of range
except IndexError:
    print("Error: List index is out of range.")
```
### **Output:**
```
Error: List index is out of range.
```

---

### **3. `KeyError` (Accessing an Invalid Dictionary Key)**
Occurs when trying to access a key that is not present in a dictionary.

```python
try:
    my_dict = {"name": "Aswanth", "age": 21}
    print(my_dict["city"])  # Key does not exist
except KeyError:
    print("Error: Key not found in dictionary.")
```
### **Output:**
```
Error: Key not found in dictionary.
```

---

### **4. `ValueError` (Invalid Value for a Function)**
Occurs when a function receives an argument of the correct type but an invalid value.

```python
try:
    num = int("abc")  # Invalid conversion
except ValueError:
    print("Error: Invalid input. Please enter a number.")
```
### **Output:**
```
Error: Invalid input. Please enter a number.
```

---

### **5. `TypeError` (Invalid Operation on Different Data Types)**
Occurs when an operation is performed on incompatible data types.

```python
try:
    result = "10" + 5  # String and integer addition
except TypeError:
    print("Error: Cannot add a string and an integer.")
```
### **Output:**
```
Error: Cannot add a string and an integer.
```

---

### **6. `FileNotFoundError` (Trying to Open a Non-Existing File)**
Occurs when trying to open a file that does not exist.

```python
try:
    file = open("non_existent_file.txt", "r")  # File does not exist
except FileNotFoundError:
    print("Error: The file was not found.")
```
### **Output:**
```
Error: The file was not found.
```

---

### **7. `AttributeError` (Accessing an Undefined Attribute)**
Occurs when trying to access an attribute that does not exist in an object.

```python
try:
    my_str = "Hello"
    print(my_str.uppercase())  # No such method
except AttributeError:
    print("Error: The attribute does not exist.")
```
### **Output:**
```
Error: The attribute does not exist.
```

---

### **8. `ImportError` (Failed Import of a Module)**
Occurs when trying to import a module that is not installed or does not exist.

```python
try:
    import non_existing_module  # Module does not exist
except ImportError:
    print("Error: Module not found.")
```
### **Output:**
```
Error: Module not found.
```

---

### **9. `NameError` (Using an Undefined Variable)**
Occurs when trying to use a variable that has not been defined.

```python
try:
    print(my_variable)  # my_variable is not defined
except NameError:
    print("Error: Variable is not defined.")
```
### **Output:**
```
Error: Variable is not defined.
```

---

### **10. `RuntimeError` (Generic Runtime Error)**
Occurs when an error is detected that does not fit into other categories.

```python
try:
    raise RuntimeError("This is a runtime error.")
except RuntimeError as e:
    print("Error:", e)
```
### **Output:**
```
Error: This is a runtime error.
```

---

## **Advanced Exception Handling**
### **Handling Multiple Exceptions**
```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
except ValueError:
    print("Error: Invalid input. Please enter a valid number.")
```

---

### **Using `finally` for Cleanup**
The `finally` block always executes, whether an exception occurs or not.

```python
try:
    file = open("sample.txt", "w")
    file.write("Hello, world!")
except:
    print("Error writing to file.")
finally:
    file.close()
    print("File closed.")
```
### **Output:**
```
File closed.
```
✔ Ensures resources are released.

---

### **Using `raise` to Manually Raise Exceptions**
```python
age = int(input("Enter age: "))
if age < 18:
    raise ValueError("You must be 18 or older.")
```
✔ Stops execution with a custom error message.

---

## **Conclusion**
- **Exception handling prevents program crashes** and ensures smooth execution.
- **Important elements:**
  - `try`: Code that might raise an exception.
  - `except`: Catches and handles exceptions.
  - `else`: Runs if no exception occurs.
  - `finally`: Always runs, useful for cleanup.
  - `raise`: Manually raises exceptions.
