### 1. Single Inheritance
A child class inherits from a single parent class.

```python
class Parent:
    def show(self):
        print("This is the Parent class.")

class Child(Parent):  # Child inherits from Parent
    def display(self):
        print("This is the Child class.")

# Creating an object of Child class
obj = Child()
obj.show()    # Inherited method
obj.display() # Child class method
```

### 2. Multiple Inheritance
A child class inherits from two or more parent classes.

```python
class Parent1:
    def method1(self):
        print("This is Parent1.")

class Parent2:
    def method2(self):
        print("This is Parent2.")

class Child(Parent1, Parent2):  # Inherits from Parent1 and Parent2
    def method3(self):
        print("This is the Child class.")

# Creating an object of Child class
obj = Child()
obj.method1()  # From Parent1
obj.method2()  # From Parent2
obj.method3()  # Child class method
```

### 3. Multilevel Inheritance
A child class inherits from another child class, creating a chain of inheritance.

```python
class Grandparent:
    def method1(self):
        print("This is the Grandparent class.")

class Parent(Grandparent):  # Parent inherits from Grandparent
    def method2(self):
        print("This is the Parent class.")

class Child(Parent):  # Child inherits from Parent
    def method3(self):
        print("This is the Child class.")

# Creating an object of Child class
obj = Child()
obj.method1()  # From Grandparent
obj.method2()  # From Parent
obj.method3()  # Child class method
```

### 4. Hierarchical Inheritance
Multiple child classes inherit from the same parent class.

```python
class Parent:
    def show(self):
        print("This is the Parent class.")

class Child1(Parent):  # Child1 inherits from Parent
    def display1(self):
        print("This is Child1.")

class Child2(Parent):  # Child2 inherits from Parent
    def display2(self):
        print("This is Child2.")

# Creating objects of both Child classes
obj1 = Child1()
obj2 = Child2()

obj1.show()    # Inherited method
obj1.display1()  # Child1 method

obj2.show()    # Inherited method
obj2.display2()  # Child2 method
```

### 5. Hybrid Inheritance
A combination of two or more types of inheritance.

```python
class A:
    def methodA(self):
        print("This is class A.")

class B(A):  # Single Inheritance
    def methodB(self):
        print("This is class B.")

class C(A):  # Hierarchical Inheritance
    def methodC(self):
        print("This is class C.")

class D(B, C):  # Multiple Inheritance
    def methodD(self):
        print("This is class D.")

# Creating an object of class D
obj = D()
obj.methodA()  # Inherited from A
obj.methodB()  # Inherited from B
obj.methodC()  # Inherited from C
obj.methodD()  # Own method
```
