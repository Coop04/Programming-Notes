### What is a Destructor?
A destructor is a special method in Python, __del__(), that is automatically called when an object is deleted or goes out of scope. It is used to release resources like closing files, freeing memory, or disconnecting from a database.

```python
class Example:
    def __init__(self, name):
        self.name = name
        print(f"Object {self.name} is created.")

    def __del__(self):
        print(f"Object {self.name} is destroyed.")

# Creating an object
obj1 = Example("Aswanth")

# Deleting the object explicitly
del obj1  

# Output will automatically appear when the program ends if not deleted explicitly.
```
