### Constructor

A constructor in Python is a special method that is automatically called when you create a new object of a class. Its main job is to initialize (i.e., set up) the object’s data.

```python
class Student:
    def __init__(self, name):
        # This is the constructor
        self.name = name
        print(f"Constructor called! Student name is {self.name}")

# Creating an object
s1 = Student("Aswanth")
```

### What is a Destructor?
A destructor is a special method in Python, __del__(), that is automatically called when an object is deleted or goes out of scope. It is used to release resources like closing files, freeing memory, or disconnecting from a database.

```python
class Example:
    def __init__(self, name):
        self.name = name
        print(f"Object {self.name} is created.")

    def __del__(self):
        print(f"Object {self.name} is destroyed.")

# Creating an object
obj1 = Example("Aswanth")

# Deleting the object explicitly
del obj1  

# Output will automatically appear when the program ends if not deleted explicitly.
```

### Polymorphism

Polymorphism in Python means that the same function or method name can be used for different types (classes), and it behaves differently depending on the object that calls it.

```python
class Animal:
    def speak(self):
        return "Animal speaks"

class Dog(Animal):
    def speak(self): # method to be overloaded must have the same name and no. of parameters as the parent class method
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

# Using polymorphism
def make_animal_speak(animal):
    print(animal.speak())

dog = Dog()
cat = Cat()

make_animal_speak(dog)  # Output: Woof!
make_animal_speak(cat)  # Output: Meow!
```

### Encapsulation

Encapsulation in Python is the concept of restricting direct access to some of an object’s components, which helps protect the internal state of the object and only exposes what’s necessary through methods.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # private attribute (private variables can only be accessed from functions defined in that particular class)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ₹{amount}")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew ₹{amount}")
        else:
            print("Insufficient funds or invalid amount")

    def get_balance(self):
        return self.__balance

# Usage
account = BankAccount("Aswanth", 1000)
account.deposit(500)
account.withdraw(300)
print("Current balance:", account.get_balance())

# Try accessing private variable directly (not recommended)
# print(account.__balance)  # This will raise an AttributeError

# __balance is a private attribute, which can't be accessed directly from outside the class.

# Access to it is provided via public methods like deposit(), withdraw(), and get_balance().
```

### Abstract Class 

In Python, an abstract class is a class that cannot be instantiated (no object can be created) and is meant to be inherited by other classes. It often contains one or more abstract methods, which are methods declared but not implemented in the abstract class — the child class must implement them.

Generally abstract methods defined in abstract class don’t have anybody but it 
is possible to have abstract methods with implementation in abstract class. 

Abstract classes in Python can have non-abstract methods.

In Python abstract class is created by deriving from the meta class ABC which 
belongs to the abc (Abstract Base Class) module

```python
from abc import ABC, abstractmethod

# Abstract class
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

# Concrete class (must implement sound())
class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

# Trying to instantiate Animal will raise an error
# a = Animal()  ❌ TypeError: Can't instantiate abstract class

# Instantiate child classes
d = Dog()
print(d.sound())  # Output: Bark

c = Cat()
print(c.sound())  # Output: Meow
```

```python
# Abstract classes in Python can have non-abstract methods.

from abc import ABC, abstractmethod

class Vehicle(ABC):
    
    @abstractmethod
    def start_engine(self):
        pass  # Must be implemented by child classes
    
    def fuel_type(self):
        return "Petrol"  # Non-abstract method

class Car(Vehicle):
    def start_engine(self):
        return "Engine started with key"

# Instantiate Car
c = Car()
print(c.start_engine())  # Output: Engine started with key
print(c.fuel_type())     # Output: Petrol
```

### What are Accessors and Mutators?

Accessors (Getters): Methods used to access (read) the value of private variables.

Mutators (Setters): Methods used to modify (write) the value of private variables.

These are especially useful when you're using encapsulation (i.e., when variables are made private using __).

```python
class Student:
    def __init__(self, name, age):
        self.__name = name      # private attribute
        self.__age = age        # private attribute

    # Accessor method (getter)
    def get_age(self):
        return self.__age

    # Mutator method (setter)
    def set_age(self, new_age):
        if new_age > 0:
            self.__age = new_age
        else:
            print("Invalid age")

# Create an object
s1 = Student("Aswanth", 20)

# Access the private age using the accessor
print("Age:", s1.get_age())  # Output: Age: 20

# Change the age using the mutator
s1.set_age(21)
print("Updated Age:", s1.get_age())  # Output: Updated Age: 21

# Try to set invalid age
s1.set_age(-5)  # Output: Invalid age
```

### Operator Overloading

Operator overloading allows you to define custom behavior for operators (like +, -, *, etc.) in your own classes. By overriding certain special methods (also known as magic methods or dunder methods), you can specify how operators should behave when applied to instances of your class.

```python
# Example that shows how to overload the + operator to add two Point objects:

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the + operator
    def __add__(self, other):
        if isinstance(other, Point):    # isinstance(other, Point) checks if other is an object of class Point (or a subclass of Point).
                                        # This is useful when you want to perform operations only between objects of the same type, like ensuring you’re adding two Point objects together.
            return Point(self.x + other.x, self.y + other.y)
        return NotImplemented

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

# Create two Point objects
p1 = Point(1, 2)
p2 = Point(3, 4)

# Use the overloaded + operator
p3 = p1 + p2  # This calls p1.__add__(p2)
print(p3)  # Output: Point(4, 6)
```

```python
# Example of Overloading Comparison Operators:

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def __eq__(self, other):
        if isinstance(other, Rectangle):
            return self.width * self.height == other.width * other.height
        return False

# Create two Rectangle objects
rect1 = Rectangle(4, 5)
rect2 = Rectangle(5, 4)

# Use the overloaded == operator
print(rect1 == rect2)  # Output: True, since the areas are the same
```

### Method Overloading

In Python, method overloading refers to defining multiple methods with the same name but different parameters. However, Python doesn't directly support method overloading in the traditional sense (like in Java or C++). Instead, Python uses default arguments or variable-length arguments (*args and **kwargs) to achieve similar functionality.

```python
class Printer:
    def print_message(self, message="Default message"):
        print(message)

# Create an instance of Printer
printer = Printer()

# Call method with no arguments
printer.print_message()  # Output: Default message

# Call method with a custom message
printer.print_message("Hello, World!")  # Output: Hello, World!
```

```python
class Area: 
    def find_area(self, a=None, b=None): 
        if a != None and b != None: 
            print("Rectangle:", (a * b)) 
        elif a != None: 
            print("square:", (a * a)) 
        else: 
            print("No figure assigned") 
obj1=Area() 
obj1.find_area() 
obj1.find_area(10) 
obj1.find_area(10,20) 
```

```python
# Example of Method Overloading Using *args:

class Calculator:
    def add(self, *args):
        return sum(args)

# Create an instance of Calculator
calc = Calculator()

# Call method with two arguments
print(calc.add(2, 3))  # Output: 5

# Call method with three arguments
print(calc.add(2, 3, 4))  # Output: 9

# Call method with four arguments
print(calc.add(1, 2, 3, 4))  # Output: 10
```

```python
# Example of Method Overloading Using **kwargs (Keyword Arguments):

class Person:
    def greet(self, **kwargs):
        if 'name' in kwargs:
            print(f"Hello, {kwargs['name']}!")
        else:
            print("Hello, Guest!")

# Create an instance of Person
person = Person()

# Call method with no keyword arguments
person.greet()  # Output: Hello, Guest!

# Call method with a keyword argument
person.greet(name="John")  # Output: Hello, John!
```
