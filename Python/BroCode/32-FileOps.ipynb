```python
# merging 2 files
with open('file1.txt','r') as f1, open('file2.txt') as f2, open('merged.txt','w') as out:
    out.write(f1.read()+'\n'+f2.read())

with open('merged.txt') as out:
    print(out.read())
```

```python
# reading file one line at a time
# efficient (Reads one line at a time without loading the entire file into memory.)
with open('file2.txt', 'r') as f:
    for line in f:
        print(line.strip())  # Removes extra newline characters
```

```python
# to read file with readlines()
# Downside: Loads the entire file into memory, which is inefficient for large files.
with open('file2.txt') as f:
    lines = f.readlines()
    for line in lines:
        print(line.strip())
    
```

```python
# to read file line by line (one line at a time)
# you don't need to use .readline() in a for loop because the for loop automatically reads the file line by line.
with open('file2.txt') as f:
    line = f.readline() # reads a single line from file without loading the whole file into the memory
    while(line):
        print(line.strip())
        line = f.readline()
```

```python
# reading file line by line and omitting line if start symbol is $
with open('file2.txt') as f:
    for line in f:
        if(line[0] == '$'):
            continue
        print(line.strip())
```
