```python
# Odd Even Function
def OddEven(num):
    if(num % 2 == 0):
        return 'Even'
    else:
        return 'Odd'
    
print(OddEven(10))
```

```python
def isPalindrome(str):
    # rev = str[::-1]
    rev = ''
    for i in str:
        rev = i + rev
    if(str == rev):
        return True
    else:
        return False
    
print(isPalindrome('amma'))
```

```python
def gcd(a, b):
    if b == 0:
        return a  # Base case: If second number is 0, return first number
    return gcd(b, a % b)  # Recursive call

# Example Usage
print(gcd(18, 12))  # Output: 6
print(gcd(10, 11))  
```

```python
# 🔥 Pure Functions vs. Modified Functions in Python 🔥
# In Python, functions can be categorized as pure functions and modified (impure) functions based on whether they modify external state or not.

# 🔹 1. Pure Functions
# A pure function: ✅ Always returns the same output for the same input.
# ✅ Does not modify any external/global variables.
# ✅ Does not cause side effects like modifying lists, dictionaries, or files.

# ✅ Example of a Pure Function:
# python
# Copy
# Edit
# def add(a, b):
#     return a + b  # No external modification, only returns a value

# print(add(3, 5))  # Output: 8
# print(add(3, 5))  # Output: 8 (same input → same output)
# 🔹 Properties of Pure Functions:
# Deterministic (output depends only on input)
# No side effects (doesn't modify external/global variables)
# Easier to test and debug (since no hidden state changes)
# 🔹 2. Modified (Impure) Functions
# A modified (impure) function:
# ❌ Modifies external variables, lists, dictionaries, etc.
# ❌ Has side effects like modifying files or printing output.
# ❌ Might not return the same result every time for the same input.

# ❌ Example of an Impure Function:
# python
# Copy
# Edit
# numbers = [1, 2, 3]

# def add_to_list(num):
#     numbers.append(num)  # Modifies external list
#     return numbers

# print(add_to_list(4))  # Output: [1, 2, 3, 4]
# print(add_to_list(5))  # Output: [1, 2, 3, 4, 5] (list changed)
# 🔹 Properties of Impure Functions:
# Modifies external state (affects global variables, lists, or files)
# Has side effects (prints output, modifies files, or changes variables)
# Harder to test/debug (because of changing state)
# 🔥 Key Differences:
# Feature	Pure Function	Modified (Impure) Function
# Depends only on input?	✅ Yes	❌ No (modifies external state)
# Side effects?	❌ No	✅ Yes (modifies variables, lists, files, etc.)
# Deterministic?	✅ Yes (same input → same output)	❌ No (output may change)
# Example	def add(a, b): return a + b	numbers.append(4)
# 🔥 Conclusion:
# Use pure functions when you want predictable, reusable, and testable code.
# Use modified functions when you need to change external data, like updating a database or modifying a file.
# Would you like an example of a real-world use case of pure vs. impure functions? 🚀
```

```python
class Rectangle:
    def __init__(self,height,width,corner_x,corner_y):
        self.height = height
        self.width = width
        self.corner_x = corner_x
        self.corner_y = corner_y

    def area(self):
        return self.height*self.width
    
    def perimeter(self):
        return 2 * (self.height + self.width)

    def center(self):
        center_x = self.corner_x + (self.width / 2)
        center_y = self.corner_y + (self.height / 2)
        return (center_x, center_y)

rect = Rectangle(10,20,5,5)
print(rect.area(),rect.perimeter(),rect.center())
```

```python
class Circle:
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius**2
    def perimeter(self):
        return 2* 3.12 * self.radius
    
cir = Circle(10)
print(cir.area(),cir.perimeter())
```
