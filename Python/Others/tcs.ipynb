```python
# Reversing a number

a = 1024
a = int(str(a)[::-1])
print(a)
# ---------------------
b = 1024
rev = 0
while(b!=0):
    rem = b%10
    rev = rev * 10 + rem
    b = b//10
print(rev)
```

```python
# Amstrong number

a = 153
temp = a
n = len(str(a))
sum = 0
while(temp!=0):
    rem = temp%10
    sum += rem**n
    temp//=10
if sum == a:
    print("Amstrong")
else:
    print("Not Amstrong")
```

```python
# Vowel Count

word = "abcdeogU"
k = 0
vowels = "aeiouAEIOU"
for i in word:
    if i in vowels:
        k+=1
print(k)
```

```python
# Largest and second largest element
l = [4,6,2,1,9,5,7]
l.sort()
largest = l[-1]
secondL = l[-2]
print(largest,secondL)

# for largest we can do 
print(max(l))
```

```python
# Removing duplicates
l = [2,4,5,5,3,2]
newL = list(set(l))
print(newL)

# we can do it by preserving the order usind dictionary
newL1 = list(dict.fromkeys(l))
print(newL1)
```

```python
# pyramid pattern

n = 4
for i in range(n):
    # for j in range():
    print(" "*(n-i-1) + "*"*(2*i+1))
```

```python
# left pyramid

n = 3
for i in range(1,n+1):
    print("*"*i)
```

```python
# right pyramid
n = 3
for i in range(1,n+1):
    print(" "*(n-i)+"*"*i)
```

```python
# fibonacci series

n = 6

a = 0
b = 1
for i in range(n+1):
    print(a,end=" ")
    a, b = b, a+b

# -------------------------
print()

# with recursion
def fib(n):
    if n == 1:
        return 1
    elif n == 0:
        return 0
    else:
        return fib(n-1)+fib(n-2)
print(fib(6))
```

```python
# Prime Check

n = 1
flag = 0
if n<2:
    print("not prime")
else:
    for i in range(2,n):
        if n%i==0:
            flag+=1
    if flag == 0:
        print("prime")
    else:
        print("not prime")
```

```python
# Factorial

n = 3
fact = 1
for i in range(1,n+1):
    fact*=i
print(fact)

# with recurtion

def fac(m):
    f = m
    if m == 1 or m==0:
        return 1
    else:
        return f * fac(m-1)
print(fac(3))
```
