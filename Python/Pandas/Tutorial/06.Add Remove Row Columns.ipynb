```python
people = {
    'first': ['Corey', 'Jane', 'John'], 
    'last': ['Schafer', 'Doe', 'Doe'], 
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['first'] + ' ' + df['last']
```

## Adding Column

```python
df['full_name'] = df['first'] + ' ' + df['last'] # cant use dot notation to create new column
df
```

```python
# same as above (using apply)
df['name'] = df.apply(lambda x: x['first'] + ' ' + x['last'], axis='columns')
df
```

## Removing Column

```python
df.drop(columns=['first','last','name'], inplace = True)
```

```python
df
```

## Splitting a string

```python
df['full_name'].str.split(' ', expand=True) # expand=True will return a dataframe instead of a series
```

```python
df[['first','last']] = df['full_name'].str.split(' ', expand=True)
df
```

## Adding and removing rows of data

adding a single row

```python
new_row = pd.DataFrame([{'first':'Tony'}])
df = pd.concat([df,new_row], ignore_index=True) # ignore_index=True will ignore the index of the new row
df 
# NaN : Not a Number (missing value)
```

## Concatenating other data frames

```python
people = {
    'first': ['Tony', 'Steve'], 
    'last': ['Stark', 'Rogers'], 
    'email': ['Ironman@email.com', 'Cap@email.com']
}
```

```python
df2 = pd.DataFrame(people)
```

```python
df2
```

```python
df =pd.concat([df,df2], ignore_index=True)
```

```python
df
```

## Removing rows

```python
df.drop(index=3, inplace=True)
df
```

```python
# using filter method
filt = df['last'] == 'Doe'
df.drop(index=df[filt].index)
```

```python
# using filter method (same as above)
df.drop(index=df[df['last']=='Doe'].index)
```
