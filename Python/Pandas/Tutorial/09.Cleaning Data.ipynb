```python
import pandas as pd
import numpy as np
```

```python
people = {
    'first': ['Corey', 'Jane', 'John', 'Chris', np.nan, None, 'NA'], 
    'last': ['Schafer', 'Doe', 'Doe', 'Schafer', np.nan, np.nan, 'Missing'], 
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com', None, np.nan, 'Anonymous@email.com', 'NA'],
    'age': ['33', '55', '63', '36', None, None, 'Missing']
}
```

```python
df = pd.DataFrame(people)
df
```

## Handling NaN values

```python
df.dropna()
```

```python
# in the background the dropna() method has
df.dropna(axis='index', how='any')
# axis can be index or columns 
# index: tell pandas to drop rows with na values
# columns: tell pandas to drop columns with na values

# how is the criteria used for dropping a row or column. how can be any or all
# any: drop row or column if any of the values are na
# all: drop row or column if all the values are na
```

```python
df.dropna(axis='columns')
```

```python
# to drop specific rows or columns with na values
df.dropna(axis='index', how='any', subset=['email']) 
# here since subset is only one column the value of how is not important, since it will have the same result 
```

```python
# Q: if there is last or email, then don't drop the row
df.dropna(axis='index', how='all', subset=['email', 'last']) 
```

## Handling Custom Missing Values

```python
df = pd.DataFrame(people)
df.replace('NA', np.nan, inplace=True)
df.replace('Missing', np.nan, inplace=True)
df
```

```python
df.dropna()
```

```python
df.dropna(how='all', subset=['email', 'last'])
```

```python
df.isna() # gives a mask of the dataframe showing whether or not they are classified as na values
```

```python
# to fill na values with a specific value
df.fillna('MISSING')
```

```python
df.fillna(0) # use inplace=True to make the changes permanent
```

## Q: To find the average of the age column

```python
# to find the average of the age column
# the age column might be a string, so we need to check the datatype first
df.dtypes
# object means it is a string datatype or a mix of datatypes
```

```python
# df['age'].mean() # this will give an error since the datatype is object
# when we have nan values in a column that we are trying to convert to numbers, then we need to use float datatype
# that is because nan values are float values
type(np.nan) # float
```

```python
df.dtypes
```

```python
df['age'] = df['age'].astype(float) # float is used because there are nan values in the column
df['age'].mean()
# another way is to fill the na values with a specific value using fillna() method
```
