### Basic Graphs

```python
x = [0,1,2,3,4,]
y = [0,2,4,6,8]

# to resize the graph
plt.figure(figsize=(5,3), dpi=100) # use dpi to specify the pixels per inch (higher dpi, higher quality)

plt.plot(x,y, label='2x', color='red', linewidth=2,linestyle='--', marker='.',markersize=10,markeredgecolor='black')
plt.title('First Graph', fontdict={'fontname': 'Comic Sans MS', 'fontsize': '20'})
plt.xlabel('X axis')
plt.ylabel('Y axis')

# to show only the interger valus along the axis (scale of your graph)
plt.xticks([0,1,2,3])
plt.yticks([0,2,4,7,8])

# to plot the legend (line names and stuff)
plt.legend()

# to save the graph
plt.savefig('mygraph.png', dpi=300) # we can use different dpi values used above in plt.figure when saving

plt.show()
```

```python
# Simple sin wave
import matplotlib.pyplot as plt
import numpy as np
import math

x = np.arange(0,math.pi*2,0.05) 
y = np.sin(x)
plt.plot(x,y)
plt.title('sine wave')
plt.xlabel('angle')
plt.ylabel('sine')
plt.show()
```

```python
# Matplotlib PyLab module
# PyLab is a procedural interface to the Matplotlib object-oriented plotting library. 
# PyLab is a convenience module that bulk imports matplotlib.pyplot (for plotting) and NumPy (for Mathematics and working with arrays) in a single  name space. 

from numpy import * 
from pylab import * 
x = linspace(-3, 3, 30) 
y = x**2 
plot(x, y) 
show() 
```

```python
from pylab import * 
x = np.arange(0, math.pi*2, 0.05) 
plot(x, sin(x)) 
plot(x, cos(x), 'r-') 
plot(x, -sin(x), 'g--') 
show() 
```

```python
# Adding Grids and Legend to the Plot 
from pylab import * 
x = np.arange(0, math.pi*2, 0.05) 
plot(x, sin(x),label='sin') 
plot(x, cos(x), 'r-',label='cos') 
plot(x, -sin(x), 'g--',label='-sin') 
grid(True) 
title('waves') 
xlabel('x') 
ylabel('sin cos -sin') 
legend(loc='upper right') 
show() 
```

```python
# Subplots with in the same plot 
# subplot(nrows, ncols, index)
# nrows (2):
# The number of rows in the grid of subplots.
# In this case, there are 2 rows.

# ncols (2):
# The number of columns in the grid of subplots.
# In this case, there are 2 columns.

# index (1):
# The position of the current subplot in the grid, numbered row-wise from left to right, starting at 1.
# In this case, 1 refers to the first subplot (top-left).

from pylab import * 
x = np.arange(0, math.pi*2, 0.05)

subplot(2,2,1) # subplot(nrows, ncols, index)
plot(x, sin(x),label='sin') 
xlabel('x') 
ylabel('sin') 
legend(loc='upper right') 
grid(True) 

subplot(2,2,2) 
plot(x, cos(x), 'r-',label='cos') 
xlabel('x') 
ylabel('cos') 
legend(loc='upper right') 
grid(True) 

subplot(2,2,3) 
xlabel('x') 
ylabel('-sin') 
plot(x, -sin(x), 'g--',label='-sin') 
legend(loc='upper right') 
grid(True)

subplot(2,2,4) 
xlabel('x') 
ylabel('tan') 
plot(x, tan(x), 'y-',label='tan') 
legend(loc='upper right') 
grid(True)
 
show()
```

```python
# program shows the default ticks and customized ticks 

import matplotlib.pyplot as plt 
import numpy as np 
 # values of x and y axes 
x = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50] 
y = [1, 4, 3, 2, 7, 6, 9, 8, 10, 5] 
figure(1)
plt.plot(x, y, 'b') 
plt.xlabel('x') 
plt.ylabel('y') 

figure(2) 
plt.plot(x, y, 'r') 
plt.xlabel('x') 
plt.ylabel('y')

# 0 is the initial value, 51 is the final value 
# (last value is not taken) and 5 is the difference 
# of values between two consecutive ticks 

plt.xticks(np.arange(0, 51, 5)) 
plt.yticks(np.arange(0, 11, 1)) 
plt.tick_params(axis='y',colors='red',rotation=45) 
plt.show() 
 
```
